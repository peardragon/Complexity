# Volume-corrected local entropy derivative figure

This notebook recreates the figure used for the abstract: the radial derivative of the volume-corrected local entropy around a reference solution.

It reads `dphi_dr_by_beta_radius.csv` from `raw_outputs` and saves the resulting figure under the corresponding `figures/18_beta_cell_90_dataset_30_reference/...` directory.

````python
from pathlib import Path

import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

matplotlib.rcParams.update({
    "figure.dpi": 120,
    "savefig.dpi": 220,
    "font.size": 13,
    "axes.titlesize": 20,
    "axes.labelsize": 16,
    "xtick.labelsize": 13,
    "ytick.labelsize": 13,
    "legend.fontsize": 12,
})
````

````python
PROJECT_ROOT = Path("/home/bjyong/Complexity/local_project/02_dnn/05_proxy_local_entropy")
RUN_DIR = Path("18_beta_cell_90_dataset_30_reference/d_0.01_to_2.50_dense")

csv_path = PROJECT_ROOT / "raw_outputs" / RUN_DIR / "summary_tables/dphi_dr_by_beta_radius.csv"
fig_dir = PROJECT_ROOT / "figures" / RUN_DIR
out_path = fig_dir / "dphi_energy_dr_by_distance_volume_corrected_local_entropy.png"

csv_path, out_path
````

````python
df = pd.read_csv(csv_path)
plot_df = (
    df[["beta", "radius", "dphi_energy_dr"]]
    .replace([np.inf, -np.inf], np.nan)
    .dropna()
    .sort_values(["beta", "radius"])
)

plot_df.head()
````

````python
fig, ax = plt.subplots(figsize=(10.2, 6.2), constrained_layout=True)

cmap = plt.get_cmap("viridis")
betas = sorted(plot_df["beta"].unique())

for index, beta in enumerate(betas):
    beta_df = plot_df.loc[plot_df["beta"] == beta]
    color = cmap(index / max(len(betas) - 1, 1))
    ax.plot(
        beta_df["radius"],
        beta_df["dphi_energy_dr"],
        marker="o",
        markersize=4.2,
        linewidth=2.0,
        color=color,
        label=fr"$\beta = {beta:g}$",
    )

ax.set_title(
    r"Radial derivative of volume-corrected local entropy" + "\n"
    + r"$\Phi_\beta^{\mathrm{corr}}(D)=\Phi_\beta(D)-\Phi_{\mathrm{vol}}(D)$",
    pad=13,
)
ax.set_xlabel(r"Distance from reference solution, $D$")
ax.set_ylabel(r"$\partial_D \Phi_\beta^{\mathrm{corr}}(D)$")
ax.grid(True, alpha=0.25, linewidth=0.8)
ax.axhline(0.0, color="black", linewidth=0.9, alpha=0.45)
ax.set_xlim(-0.02, 2.62)
ax.legend(loc="center left", bbox_to_anchor=(1.02, 0.5), frameon=False, handlelength=2.4)

fig_dir.mkdir(parents=True, exist_ok=True)
fig.savefig(out_path)
out_path
````
